```python
## CREATING SERIES
# there are multiple ways to create a series in pandas.
```

```python
import numpy as np
import pandas as pd
```

```python
labels = ['a', 'b', 'c']
mylist = [10, 20, 30]
arr = np.array([10, 20, 30])
d = {'abs': 10, 'b': 20, 'c': 30}
```

```python
pd.Series(mylist)
```

```python
pd.Series(mylist,index = labels)
```

```python
pd.Series(arr)
```

```python
pd.Series(d)
```
